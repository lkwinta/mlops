```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("sentence-transformers/clip-ViT-B-32")
```

```python
from PIL import Image
import matplotlib.pyplot as plt

img_path = "./data/images/small/f6/f6ae8ed8.jpg"

img = Image.open(img_path)

plt.imshow(img)
plt.axis("off")
plt.show()
```

```python
img_emb = model.encode(img)
img_emb.shape
```

```python
text_emb = model.encode(["telephone", "laptop", "PC", "dog", "mobile"])
text_emb.shape
```

```python
similarity_scores = model.similarity(img_emb, text_emb)
print(similarity_scores)
```

It is a multimodal CLIP model (Contrastive Language–Image Pre-training) 
that maps images and text into the same vector space, allowing their semantic similarity to be compared

What is the vector size produced by clip-ViT-B-32?

- The vector size produced by clip-ViT-B-32 is 512.

What do ViT, B and 32 mean in the model name?

- ViT stands for Vision Transformer, which is the architecture used for the model.
- B indicates the model size, specifically the "Base" variant.
- 32 refers to the patch size used in the Vision Transformer, meaning each image is divided into 32x32 patches.
